## W4A22

## W4A25
Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1.
Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập 


```python
so_lon_nhat = None
so_nho_nhat = None
while True:
    s = int(input())
    if s == -1:
        break
    if so_lon_nhat is None or s > so_lon_nhat:
        so_lon_nhat = s
    if so_nho_nhat is None or s < so_nho_nhat:
        so_nho_nhat = s
if so_lon_nhat is None:
    print("Không có dữ liệu")
else:
    print("Số lớn nhất:", so_lon_nhat)
    print("Số nhỏ nhất:", so_nho_nhat)
```

## W4A26
[Fibonaci1]
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
bằng tổng hai số trước đó
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
quá A

```python
A=int(input())
s=1
b=1
while b<=A:
    c=s+b
    s=b
    b=c
print(s)

```

## W4A27
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có
ký tự đặc biệt, không số, không có dấu câu, chỉ có ký tự chữ và khoảng trắng)
Dùng cách: for trên index; KHÔNG dùng hàm split(), len()


```python
chuoi = input()
dem = 0
dang_o_trong_tu = False
for ky_tu in chuoi:
    if ky_tu != ' ':
        if not dang_o_trong_tu:
            dem += 1
            dang_o_trong_tu = True
    else:
        dang_o_trong_tu = False
print("Số từ là:", dem)
```

## W4A28
Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi
Dùng cách: for trên index; KHÔNG dùng hàm split()

```python
a = input()
i = 0
while i < len(a) and a[i] == ' ':
    i += 1
while i < len(a) and a[i] != ' ':
    print(a[i], end="")
    i += 1
```

## W4A29
Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành thường và thường thành hoa.


```python
chuoi = input()
ket_qua = ""
for ky_tu in chuoi:
    if ky_tu.islower():
        ket_qua += ky_tu.upper()
    else:
        ket_qua += ky_tu.lower()
print(ket_qua)
```

## W4A31
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in
thường, bao nhiêu ký tự số 

```python
a=input()
b=0
c=0 
d=0
for i in a:
    if i.isupper():
        b+=1
    elif i.islower():
        c+=1
    elif i.isdigit():
        d+=1
print(f"hoa={b}, thường={c}, số={d}")
```
