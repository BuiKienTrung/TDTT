## W6A16
Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n = int(input())
m = int(input())

a = []

for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

for i in range(n):
    for j in range(m):
        print(f"{a[i][j]:>4}", end="")
    print()
```

## W6A17
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảnh

```python
n = int(input())

a = []

for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

for i in range(n):
    print(a[i][i], end=" ")

print()

for i in range(n):
    print(a[i][n - 1 - i], end=" ")
```

## W6A18
Nhâp vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
n, m, k = map(int, input().split())

a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

tong_cot = 0
for i in range(n):
    tong_cot += a[i][k]

tong_hang = 0
for j in range(m):
    tong_hang += a[k][j]

print(tong_cot)
print(tong_hang)
```

## W6A19
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list

```python
def vitri_max(a):
    max_value = a[0]
    vitri = 0
    for i in range(1, len(a)):
        if a[i] > max_value:
            max_value = a[i]
            vitri = i
    return vitri
a=list(map(int,input().split()))
print(vitri_max(a))
```

## W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1

```python
def vitri(a,k):
    for i in range(len(a)):
        if a[i]==k:
            return i
    return -1
a=list(map(int,input().split()))
k=int(input())
print(vitri(a,k))

```

## W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện


```python
a = input().split()
d = {}
for i in range(len(a)):
    if a[i] not in d:
        d[a[i]] = (i,)
    else:
        d[a[i]] = d[a[i]] + (i,)
print(d)
```

## W6A23
 VIết hàm xóa đi các số <x có trong mảng

```python
def xoa_nho_hon_x(a, x):
    for i in a[:]:
        if i < x:
            a.remove(i)
    return a
a=list(map(int,input().split()))
x=int(input())
print(xoa_nho_hon_x(a, x))
```

# W6A24
Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def day_0_ve_dau(a):
    so0=[]
    khac0=[]
    for i in a:
        if i==0:
            so0.append(i)
        else:
            khac0.append(i)
    return so0+khac0
a=list(map(int,input().split()))
print(day_0_ve_dau(a))
```

# W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối


```python
a = list(map(int, input().split()))
tong = 0
for i in range(1, len(a) - 1):
    if a[i] > a[i - 1] and a[i] > a[i + 1]:
        tong += a[i]
print(tong)
```

# W6A27
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). nếu không có in ra -1


```python
a=list(map(int,input().split()))
x=int(input())
vitri=[]
for i in range(len(a)):
    if a[i]<x:
        vitri.append(i)
if len(vitri)==0:
    print(-1)
else:
    print(vitri)
```

# W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
a=list(map(int,input().split()))
tang=True
giam=False
for i in range(len(a)-1):
    if a[i]>=a[i+1]:
        tang=False
    elif a[i]<=a[i+1]:
        giam=False
if tang:
    print("Tang")
elif giam:
    print("Giam")
else:
    print("Vo danh")
```
