## Bài 3 : 
Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.

```python
import math
a, b, c = map(float, input().split())
if a + b <= c or a + c <= b or b + c <= a:
    print("Khong la ba canh cua mot tam giac")
else:
    y = a + b + c
    p = y / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    if a == b == c:
        t = "Tam giác đều"
    elif a == b or b == c or a == c:
        if a*a + b*b == c*c or a*a + c*c == b*b or b*b + c*c == a*a:
            t = "Tam giác vuông cân"
        else:
            t = "Tam giác cân"
    elif a*a + b*b == c*c or a*a + c*c == b*b or b*b + c*c == a*a:
        t = "Tam giác vuông"
    else:
        t = "Tam giác thường"
    print(t)
    print(f"Chu vi: {y}")
    print(f"Diện tích: {s}")
```

## Bài 5: 
Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python

import math
a, b, c = map(int, input().split())
if a == 0:
    if b == 0:
        if c == 0:
            print("Phương trình vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        print(f"Phương trình bậc 1 có nghiệm x = {-c/b:.2f}")
else:
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phương trình có nghiệm kép x = {x:.2f}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print(f"Phương trình có 2 nghiệm phân biệt:")
        print(f"x1 = {x1:.2f}")
        print(f"x2 = {x2:.2f}")
```

## Bài 8:
 Giải hệ phương trình tuyến tính:
ax  by  m
cx  dy  n

```python
a, b, c, d, m, n = map(float, input().split())
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c
if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"Hệ có nghiệm duy nhất: x = {x:.2f}, y = {y:.2f}")
else:
    if Dx == 0 and Dy == 0:
        print("Hệ có vô số nghiệm")
    else:
        print("Hệ vô nghiệm")
```

## Bài 9:
 Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao nhiêu giây. 

```python
x = int(input())
h = x//3600
p = (x - h*3600)//60
a = x - h*3600 - p*60
print(f"{h} giờ {p} phút {a} giây")
```

## Bài 10: 
Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
a, b, r, x, y = map(int, input().split())
if (x-a)**2+(y-b)**2<r**2:
    print("nằm trong đường tròn")
elif (x-a)**2+(y-b)**2==r**2:
    print("thuộc đđường tròn")
else:
    print("nằm ngoài đường tròn")
```
