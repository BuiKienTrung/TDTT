## W5A16
Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x


```python
def is_armstrong(x):
    n = abs(x)  
    k = len(str(n)) 
    tong = 0
    temp = n
    while temp > 0:
        digit = temp % 10
        tong += digit ** k
        temp //= 10
    return tong == n
n = int(input())
if is_armstrong(n):
    print("Là số Armstrong")
else:
    print("Không phải")
```

## W5A18 
VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên

```python
def sodoixung(n):
    if (n<0):
        return False
    a=n
    b=0
    while a>0:
        s=a%10
        b=b*10+s
        a=a//10
    return b==n
l=sodoixung(121)
i=sodoixung(123)
e=sodoixung(1221)
print(l,i,e)
```

## W5A19 
VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.


```python
def sokidieu(n):
    if n<=0:
        return False
    while n!=1 and n!=4:
        tong=0
        while n>0:
            a=n%10
            tong+=a*a
            n//=10
        n=tong
    return tong==1
b=sokidieu(19)
c=sokidieu(5)
print(b,c)
```

## W5A20
 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    try:
        int(s)
        return 'int'
    except:
        try:
            float(s)
            return 'float'
        except:
            return 'neither'
a= detect_number_type("12")
b= detect_number_type("1.7")
c= detect_number_type("abc")
print(a,b,c)
```

```python
def is_one_product_of_others(a,b,c):
    if (a == b * c) or (b == a * c) or (c == a * b):
        return True
    else:
        return False
f=is_one_product_of_others(0,7,0)
g=is_one_product_of_others(1,6,7)
print(f,g)
```

## W5A24 
Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1,y1,x2,y2,x3,y3):
     if (x2 - x1)*(y3 - y1) == (x3 - x1)*(y2 - y1):
          return True
     else:
          return False
a=are_collinear(1,1,2,2,3,3)
b=are_collinear(1,2,3,4,1,3)
print(a,b)
```

## W5A25
Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    def area(A, B, C):
        return abs((A[0]*(B[1]-C[1]) + B[0]*(C[1]-A[1]) + C[0]*(A[1]-B[1])) / 2)
    A, B, C, D = quad
    S = area(A, B, C)
    S1 = area(A, B, p)
    S2 = area(B, C, p)
    S3 = area(C, A, p)
    in_ABC = abs(S - (S1 + S2 + S3)) < 1e-6
    S = area(A, C, D)
    S1 = area(A, C, p)
    S2 = area(C, D, p)
    S3 = area(D, A, p)
    in_ACD = abs(S - (S1 + S2 + S3)) < 1e-6
    return in_ABC or in_ACD
quad = [(0,0), (4,0), (4,3), (0,3)]
p = (2,1)
print(point_in_quadrilateral(quad, p))
```

Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def khac0(n):
    count=0
    for i in n:
        i=abs(i)
        if i==0:
            continue
        while i>0:
            a=i%10
            if a!=0:
                count+=1
            i//=10
    return count
n = map(int, input().split())
print(khac0(n))




```

## W5A28
 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3)

```python
def arccos(x):
    pi = 3.141592653589793
    if x >= 1:
        return 0
    if x <= -1:
        return pi
    return pi/2 - x - (x**3)/6
def goc_ABC(A, B, C):
    # vector BA và BC
    BAx = A[0] - B[0]
    BAy = A[1] - B[1]
    BCx = C[0] - B[0]
    BCy = C[1] - B[1]
    dot = BAx*BCx + BAy*BCy
    lenBA = (BAx**2 + BAy**2)**0.5
    lenBC = (BCx**2 + BCy**2)**0.5
    if lenBA == 0 or lenBC == 0:
        return 0
    cos_theta = dot / (lenBA * lenBC)
    if cos_theta > 1:
        cos_theta = 1
    if cos_theta < -1:
        cos_theta = -1
    rad = arccos(cos_theta)
    deg = rad * 180 / 3.141592653589793
    return deg
A = (1,0)
B = (0,0)
C = (0,1)
print(goc_ABC(A,B,C))
```

## W5A30
 Viết hàm tính hệ số nhị thức C(n, k)

```python
def giaithua(n):
    a=1
    for i in range (1,n+1):
        a *= i
    return a
def C(n,k):
    if (k<0 or k>n):
        return False
    return giaithua(n)//(giaithua(k)*giaithua(n-k))
print(C(5,2))

```

## W5A33
Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,giá km, phí chờ, phụ phí ban đêm và giảm giá.

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if (car_type==4):
        gia_mo_cua=12000
        gia_km=10000
    else:
        gia_mo_cua=15000
        gia_km=13000
    if (km<=2):
        gia_km = gia_mo_cua
    else:
        gia_km= gia_mo_cua + (km-2)*gia_km
    phi_cho=wait_minutes*700
    tong=phi_cho+gia_km
    if (start_hour>=22 or start_hour<=5):
        tong=tong*1.2
    return tong
print(smart_taxi_fare(4,23,7,1))




```

## W5A34
Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
    if(n<=0):
        return 0
    else:
        a=0
        for i in range(1,n+1):
            a+=i
        return a
print(sum_1_to_n(-2))
```
