# W7A16
Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
a=list(map(int,input().split()))
d={}
for x in a:
    if x in d:
        d[x]+=1
    else:
        d[x]=1
ans=0
for x in d:
    cnt=d[x]
    if cnt>x:
        ans+=cnt-x
    elif cnt<x:
        ans+=cnt
print(ans)

```

W7A22
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
d = {}
for _ in range(n):
    hoten = input()
    ten = hoten.split()[-1]
    if ten in d:
        d[ten] += 1
    else:
        d[ten] = 1
max_dem = max(d.values())
kq = []
for ten in d:
    if d[ten] == max_dem:
        kq.append(ten)
print(min(kq))
```

# W7A25
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
s = input()
k = int(input())
kq = ""
start = 0
for i in range(k):
    end = len(s) - (k - i) + 1
    mx = max(s[start:end])
    vt = s.index(mx, start, end)
    kq += mx
    start = vt + 1
print(kq)
```

# W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ

```python
s = input()

stack = []

ok = True

for ch in s:

    # ngoặc mở
    if ch in "([{":
        stack.append(ch)

    # ngoặc đóng
    else:

        if len(stack) == 0:
            ok = False
            break

        top = stack[-1]

        if ch == ')' and top != '(':
            ok = False
            break

        if ch == ']' and top != '[':
            ok = False
            break

        if ch == '}' and top != '{':
            ok = False
            break

        stack.pop()

# nếu còn ngoặc mở chưa đóng
if len(stack) != 0:
    ok = False

if ok:
    print("YES")
else:
    print("NO")
```

# W7A29
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
a = list(map(int, input().split()))

start = 0
max_start = 0

dai = 1
max_dai = 1

for i in range(1, len(a)):

    # còn tăng
    if a[i] > a[i-1]:
        dai += 1

    # bị ngắt
    else:

        if dai > max_dai:
            max_dai = dai
            max_start = start

        start = i
        dai = 1

# kiểm tra dãy cuối
if dai > max_dai:
    max_dai = dai
    max_start = start

print(*a[max_start:max_start + max_dai])
```

# W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím

```python
a = list(map(int, input().split()))

n = len(a)

dp = [1] * n
truoc = [-1] * n

for i in range(n):

    for j in range(i):

        if a[j] < a[i] and dp[j] + 1 > dp[i]:

            dp[i] = dp[j] + 1
            truoc[i] = j

# tìm vị trí có LIS dài nhất
vt = 0

for i in range(n):

    if dp[i] > dp[vt]:
        vt = i

# truy vết
kq = []

while vt != -1:

    kq.append(a[vt])

    vt = truoc[vt]

kq.reverse()

print(*kq)
```
